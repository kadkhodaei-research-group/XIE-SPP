# Synthesizability prediction

> Requires the dev installation (see README): `uv pip install -e ".[dev]"`

The input to the models can be CIF files, POSCAR (use `input_format='vasp'`), PyMatGen Structures or ASE Atoms.

## Version 2 (recommended)

```python
import warnings
warnings.filterwarnings('ignore', category=FutureWarning)  # pandas deprecation warnings

from xiespp import synthesizability, get_test_samples
# or
# from xiespp import synthesizability_2 as synthesizability

samples = sorted(get_test_samples('CSi'))
samples
```

```python
model = synthesizability.SynthesizabilityPredictor()  # device='/device:GPU:0' to use a GPU
model
```

```python
yp = model.predict(samples, verbose=0)
yp
```

Setting `return_all_ensembles=True` returns the predictions of every rotational ensemble in addition to the mean:

```python
model.predict(samples, return_all_ensembles=True, verbose=0).iloc[:, :5]
```
